# host-it: Cloud-to-Local LLM Bridge

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/berlin-566/host-it/blob/main/notebooks/colab_host_it.ipynb)

Host high-performance LLMs (DeepSeek-R1, Qwen 2.5 14B/32B, Llama 3.3 70B) on free Google Colab GPUs and seamlessly use them on your low-end local machine via `host-it`.

### Instructions:
1. **Switch to GPU**: Click **Runtime** -> **Change runtime type** -> Select **T4 GPU** (or A100/L4 if Colab Pro).
2. *(Optional)* Check the box below if you want to mount Google Drive to keep downloaded models permanently.
3. Click **Run All** (or `Ctrl+F9`).
4. Copy the `host-it connect ...` command generated below and run it in your local laptop's terminal!

```python
# @title Start host-it Cloud Runner { display-mode: "form" }
# @markdown Choose whether to save downloaded models to your Google Drive so you don't have to re-download them next time:
PERSIST_TO_GOOGLE_DRIVE = False # @param {type:"boolean"}

import os
import subprocess

if PERSIST_TO_GOOGLE_DRIVE:
    from google.colab import drive
    drive.mount('/content/drive')
    models_dir = '/content/drive/MyDrive/host_it_models'
    os.makedirs(models_dir, exist_ok=True)
    os.environ['OLLAMA_MODELS'] = models_dir
    print(f"[STORAGE] Persistent model storage enabled at: {models_dir}")

# Keepalive in background
from IPython.display import display, Javascript, HTML
display(HTML('''
<script>
function keepAlive() {
   console.log("[host-it] Heartbeat active");
}
setInterval(keepAlive, 60000);
</script>
'''))

# Fetch and run server supervisor
!curl -fsSL https://raw.githubusercontent.com/berlin-566/host-it/main/remote/bootstrap.sh | bash
```
